### Problem 01 · Easy — basic array from data
Create a 1-D integer array containing exactly `[5, 10, 15, 20, 25]`.  
Its dtype must be `int64` and its shape `(5,)`.


```python
# Solution
arr01 = np.array([5, 10, 15, 20, 25])
```

### Problem 02 · Easy — zeros with specific dtype
Create a 3×4 matrix of zeros with dtype `float32`.  
It must use exactly **48 bytes** of memory.


```python
# Solution 01
#arr02 = np.zeros((3 ,4) , dtype= np.float32)
# Solution 02 
arr02 = np.zeros((3,4))
arr02 = arr02.astype(np.float32)
```

### Problem 03 · Easy — linspace
Create an array of **9 evenly spaced values from 0 to 1 inclusive**.  
The first element must be exactly `0.0`, the last exactly `1.0`,  
and every step must be exactly `0.125`.


```python
#Solution
arr03 = np.linspace(0, 1 ,9)

print(f"The array : {arr03}")
```

```python
assert arr03 is not None
assert len(arr03) == 9
assert arr03[0]  == 0.0
assert arr03[-1] == 1.0
assert np.allclose(np.diff(arr03), 0.125)
print("✓ Problem 03 passed")
```

### Problem 05 · Easy — identity and diagonal
Create a 5×5 identity matrix.  
Then use `np.diag` on it to extract its diagonal as a separate array.  
The diagonal must equal `[1., 1., 1., 1., 1.]`.


```python
# Solution
I5 = np.eye(5)
diag5 = np.diag(I5)
```

### Problem 06 · Easy — full and full_like
Create a 4×4 matrix where every element is `7.0` (dtype `float64`).  
Then create a second matrix with the **same shape and dtype** but filled with `-1.0`,  
using `np.full_like`.


```python
# Solution
sevens = np.full((4, 4) , 7 , dtype= np.float64)
neg_ones = np.full_like(sevens, -1.0)
```

```python
assert sevens.shape   == (4, 4)
assert sevens.dtype   == np.float64
assert (sevens == 7.0).all()

assert neg_ones.shape == (4, 4)
assert neg_ones.dtype == np.float64
assert (neg_ones == -1.0).all()
print("✓ Problem 06 passed")
```

### Problem 07 · Medium — logspace
Create an array of **50 values logarithmically spaced between 1 and 100000**  
(i.e. `10^0` to `10^5`).  
Verify: first element is `1.0`, last is `100000.0`, and every consecutive  
ratio is constant (that's the definition of a geometric sequence).


```python
# Solution
arr07 = np.logspace(0, 5, num =50)
```

```python
assert arr07 is not None
assert len(arr07) == 50
assert np.isclose(arr07[0],  1.0)
assert np.isclose(arr07[-1], 100000.0)
ratios = arr07[1:] / arr07[:-1]
assert np.allclose(ratios, ratios[0], rtol=1e-5), "Ratios are not constant"
print("✓ Problem 07 passed")
```

### Problem 08 · Medium — dtype casting and memory
You are given a `float64` array of shape `(500, 200)`.  
Cast it to `float32` and to `float16`.  
Print the memory used by each, then verify:
- `float32` uses exactly half the bytes of `float64`
- `float16` uses exactly a quarter of the bytes of `float64`


```python
# Setup — do not modify
rng = np.random.default_rng(0)
arr_f64 = rng.random((500, 200))

# Solution
arr_f32 = arr_f64.astype(np.float32)
arr_f16 = arr_f64.astype(np.float16)
```

```python
assert arr_f32 is not None and arr_f16 is not None
assert arr_f32.dtype == np.float32
assert arr_f16.dtype == np.float16
assert arr_f32.shape == arr_f64.shape
assert arr_f16.shape == arr_f64.shape
assert arr_f64.nbytes == arr_f32.nbytes * 2,  "float32 should be half"
assert arr_f64.nbytes == arr_f16.nbytes * 4,  "float16 should be a quarter"
print(f"float64 : {arr_f64.nbytes/1e6:.3f} MB")
print(f"float32 : {arr_f32.nbytes/1e6:.3f} MB  (÷2)")
print(f"float16 : {arr_f16.nbytes/1e6:.3f} MB  (÷4)")
print("✓ Problem 08 passed")
```

### Problem 09 · Medium — fromfunction
Use `np.fromfunction` to create a 6×6 matrix where each element `[i, j]`  
equals `i + j` (0-indexed, dtype int).  
No loops. The top-left element must be `0`, the bottom-right must be `10`.


```python
# Solution
arr09 = np.fromfunction(lambda i ,j : i + j , (6, 6))
```

```python
assert arr09 is not None
assert arr09.shape == (6, 6)
assert arr09[0, 0]  == 0
assert arr09[5, 5]  == 10
assert arr09[2, 3]  == 5
# Verify all values
for i in range(6):
    for j in range(6):
        assert arr09[i, j] == i + j, f"Wrong at [{i},{j}]: got {arr09[i,j]}"
print("✓ Problem 09 passed")
```
