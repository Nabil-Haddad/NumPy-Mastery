### Problem 01 · Easy — element-wise arithmetic
Given two arrays `a` and `b` of the same shape, compute:
1. `total` — their element-wise sum
2. `diff` — their element-wise difference 
3. `ratio` — their element-wise division 


```python
a = np.array([10.0, 20.0, 30.0, 40.0])
b = np.array([2.0, 4.0, 5.0, 8.0])

# YOUR CODE
total = a + b
diff  = a - b
ratio = a / b
```

### Problem 03 · Easy — axis aggregation
Given `grades` (shape `(4, 3)` — 4 students, 3 exams):
1. `student_avg` — average score **per student** (one value per row) → shape `(4,)`
2. `exam_avg` — average score **per exam** (one value per column) → shape `(3,)`


```python
grades = np.array([[80, 90, 70],
                    [60, 65, 70],
                    [95, 85, 100],
                    [50, 55, 60]])

# YOUR CODE
student_avg = grades.mean(axis = 1)
exam_avg    = grades.mean(axis = 0)
```

```python
assert student_avg.shape == (4,)
assert np.allclose(student_avg, [80.0, 65.0, 93.333333, 55.0])
assert exam_avg.shape == (3,)
assert np.allclose(exam_avg, [71.25, 73.75, 75.0])
print("✓ Problem 03 passed")
```

### Problem 05 · Medium — broadcasting a row vector
Given `prices` (shape `(3, 4)` — 3 stores, 4 products) and `discount`
(shape `(4,)` — one discount rate per product), compute `final_prices`
where each product's price is reduced by its discount rate.

`final_prices[i, j] = prices[i, j] * (1 - discount[j])`


```python
prices   = np.array([[100.0, 200.0, 50.0, 80.0],
                      [120.0, 180.0, 60.0, 90.0],
                      [110.0, 190.0, 55.0, 85.0]])
discount = np.array([0.1, 0.2, 0.0, 0.5])   # 10%, 20%, 0%, 50% off

# Solution
final_prices = prices * (1 - discount)
```

```python
expected = prices * (1 - discount)
assert final_prices.shape == (3, 4)
assert np.allclose(final_prices, expected)
assert np.isclose(final_prices[0, 1], 160.0)   # 200 * 0.8
assert np.isclose(final_prices[1, 3], 45.0)    # 90 * 0.5
print("✓ Problem 05 passed")
```

### Problem 06 · Medium — broadcasting a column vector
Given `M` (shape `(4, 3)`), subtract the **row mean** from every element
in that row, so each row ends up with mean ≈ 0.

Hint: you'll need `axis=1` and `keepdims=True` together.


```python
M = np.array([[10.0, 20.0, 30.0],
              [ 5.0,  5.0,  5.0],
              [ 1.0,  2.0,  3.0],
              [ 8.0,  6.0, 10.0]])

# Solution
M_centered = M - M.mean(axis =1, keepdims = True) 
```

```python
assert M_centered.shape == M.shape
row_means_after = M_centered.mean(axis=1)
assert np.allclose(row_means_after, 0.0, atol=1e-10)
print("✓ Problem 06 passed")
```

`outer_sum[i, j] = i + j` (0-indexed), using only broadcasting  
(a `(6,1)` array plus a `(1,6)` array).

Without computing any actual array operation, use `np.broadcast_shapes`
to determine:

```python
# YOUR CODE
shape_a = np.broadcast_shapes((5, 1, 3) ,(1, 4, 1) )

try:
    np.broadcast_shapes((3, 4), (4, 5))
    is_compatible = True
except ValueError as e:
    is_compatible = False

```
