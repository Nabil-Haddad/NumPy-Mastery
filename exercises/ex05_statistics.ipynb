### Problem 01 · Easy — basic descriptive statistics
Compute the following for `data` and store each in the named variable:
- `mean_val` — arithmetic mean
- `median_val` — median
- `std_val` — **sample** standard deviation
- `range_val` — max minus min


```python
data = np.array([4.0, 8.0, 15.0, 16.0, 23.0, 42.0])

# Solution
mean_val = np.mean(data)
median_val = np.median(data)
std_val = np.std(data, ddof=1)
range_val = data.max() - data.min()
```

```python
assert np.isclose(mean_val,   18.0)
assert np.isclose(median_val, 15.5)
assert np.isclose(std_val, np.std(data, ddof=1))
assert np.isclose(range_val, 38.0)
print("✓ Problem 01 passed")
```

### Problem 02 · Easy — ddof matters
Given `sample` (8 values drawn from some population), compute:
- `pop_std` — population std (ddof=0)
- `samp_std` — sample std (ddof=1)
- `diff` — the difference between them

Verify that `samp_std > pop_std` (Bessel's correction always increases the estimate).


```python
sample = np.array([2.0, 4.0, 4.0, 4.0, 5.0, 5.0, 7.0, 9.0])

# Solution
pop_std = np.std(sample, ddof=0)
samp_std = np.std(sample, ddof=1)
diff = samp_std - pop_std
```

```python
assert np.isclose(pop_std,  np.std(sample, ddof=0))
assert np.isclose(samp_std, np.std(sample, ddof=1))
assert np.isclose(diff, samp_std - pop_std)
assert samp_std > pop_std, "Sample std must be greater than population std"
print(f"✓ Problem 02 passed — pop={pop_std:.4f}, samp={samp_std:.4f}, diff={diff:.4f}")
```

### Problem 03 · Easy — percentiles
Given `scores`, compute:
- `q1`, `q2`, `q3` — the 25th, 50th, 75th percentiles
- `iqr` — the interquartile range (Q3 - Q1)
- `p90` — the 90th percentile


```python
scores = np.array([52, 68, 71, 74, 78, 81, 85, 88, 91, 95])

# Solution
q1 = np.percentile(scores, 25)
q2 = np.percentile(scores, 50)
q3 = np.percentile(scores, 75)
iqr = q3 - q1
p90 = np.percentile(scores, 90)
```

```python
assert np.isclose(q1,  np.percentile(scores, 25))
assert np.isclose(q2,  np.percentile(scores, 50))
assert np.isclose(q3,  np.percentile(scores, 75))
assert np.isclose(iqr, q3 - q1)
assert np.isclose(p90, np.percentile(scores, 90))
print(f"✓ Problem 03 passed — Q1={q1}, Q2={q2}, Q3={q3}, IQR={iqr}, P90={p90}")
```

### Problem 05 · Medium — per-column statistics on 2-D data
Given `X` (shape `(100, 4)` — 100 samples, 4 features), compute:
- `col_means` — mean of each column → shape `(4,)`
- `col_stds` — sample std of each column → shape `(4,)`
- `col_medians` — median of each column → shape `(4,)`

All without loops.


```python
rng = np.random.default_rng(0)
X = rng.normal(loc=np.array([0, 5, -3, 10]),
               scale=np.array([1, 2, 0.5, 3]),
               size=(100, 4))

# Solution
col_means = X.mean(axis=0)
col_stds = X.std(axis=0, ddof=1)
col_medians = np.median(X, axis=0)
```

```python
assert col_means.shape == (4,)
assert col_stds.shape == (4,)
assert col_medians.shape == (4,)
assert np.allclose(col_means, X.mean(axis=0))
assert np.allclose(col_stds,X.std(axis=0, ddof=1))
assert np.allclose(col_medians, np.median(X, axis=0))
print(f"✓ Problem 05 passed")
print(f"means: {col_means.round(2)}")
print(f"stds : {col_stds.round(2)}")
```

### Problem 06 · Medium — correlation
You are given three arrays `x`, `y`, `z`. Without looking at a plot:
1. Compute `r_xy` — the Pearson correlation between `x` and `y`
2. Compute `r_xz` — the Pearson correlation between `x` and `z`
3. Store `stronger` as the string `'xy'` or `'xz'` depending on which
   pair has the stronger (higher absolute) correlation


```python
rng = np.random.default_rng(5)
x = np.arange(20, dtype=float)
y = 2 * x + rng.normal(0, 3, 20)     # strong positive correlation with x
z = rng.normal(0, 1, 20)             # no relationship with x

# Solution
r_xy = np.corrcoef(x, y)[0, 1]
r_xz = np.corrcoef(x, z)[0, 1]
stronger = 'xy' if abs(r_xy) > abs(r_xz) else 'xz'
```

```python
corr_xy = np.corrcoef(x, y)[0, 1]
corr_xz = np.corrcoef(x, z)[0, 1]
assert np.isclose(r_xy, corr_xy)
assert np.isclose(r_xz, corr_xz)
assert stronger in ('xy', 'xz')
assert stronger == ('xy' if abs(corr_xy) > abs(corr_xz) else 'xz')
print(f"✓ Problem 06 passed — r_xy={r_xy:.4f}, r_xz={r_xz:.4f}, stronger={stronger}")
```

### Problem 07 · Medium — random train/test split
Given `n=200` samples (indexed 0 to 199), use `rng.permutation` to
create a random 80/20 split:
- `train_idx` — 160 indices (first 80% of the permutation)
- `test_idx` — 40 indices (last 20%)

Verify: no index appears in both sets, and together they cover all 200.


```python
rng = np.random.default_rng(42)
n = 200

# Solution
idx = rng.permutation(n)
train_idx = idx[:160]
test_idx = idx[160:]
```

```python
assert len(train_idx) == 160
assert len(test_idx) == 40
assert len(np.intersect1d(train_idx, test_idx)) == 0, "Sets must not overlap"
assert len(np.union1d(train_idx, test_idx)) == 200,   "Together must cover all 200"
print("✓ Problem 07 passed")
```

### Problem 08 · Medium — histogram analysis
Generate 10,000 samples from a standard normal distribution.
1. Compute `counts` and `edges` using `np.histogram` with 20 bins
2. Find `peak_bin` — the index of the bin with the most counts
3. Find `peak_center` — the centre value of that bin  
   (hint: `(edges[i] + edges[i+1]) / 2`)
4. Verify the peak is near 0 (within 0.5)


```python
rng = np.random.default_rng(42)
data = rng.normal(0, 1, 10000)

# Solution
counts, edges = np.histogram(data, bins=20)
peak_bin = np.argmax(counts)
peak_center = (edges[peak_bin] + edges[peak_bin + 1]) / 2

```

```python
assert len(counts) == 20
assert len(edges)  == 21
assert peak_bin == np.argmax(counts)
expected_center = (edges[peak_bin] + edges[peak_bin+1]) / 2
assert np.isclose(peak_center, expected_center)
assert abs(peak_center) < 0.5, f"Peak at {peak_center:.3f}, expected near 0"
print(f"✓ Problem 08 passed — peak bin centre: {peak_center:.4f}")
```

### Problem 09 · Hard — Monte Carlo estimation of π
Use the following method to estimate π:
1. Generate `N=500_000` random 2-D points in the square `[-1,1] × [-1,1]`
   (two columns: x and y coordinates)
2. Compute `distances` — the Euclidean distance of each point from the origin
3. Count `n_inside` — points where distance ≤ 1 (i.e. inside the unit circle)
4. Estimate `pi_estimate = 4 * n_inside / N`
5. Verify your estimate is within 0.01 of the true π

No loops.


```python
rng = np.random.default_rng(42)
N = 500_000

# Solution
points = rng.uniform(-1, 1, size=(N, 2))          # shape (N, 2)
distances = np.sqrt((points ** 2).sum(axis=1))        # shape (N,)
n_inside = (distances <= 1).sum()
pi_estimate = 4 * n_inside / N
```

```python
assert points.shape == (N, 2)
assert distances.shape == (N,)
assert 0 < n_inside < N
assert abs(pi_estimate - np.pi) < 0.01, f"Got {pi_estimate}, expected ~{np.pi:.5f}"
print(f"✓ Problem 09 passed — π estimate: {pi_estimate:.5f} (error: {abs(pi_estimate-np.pi):.5f})")
```

### Problem 10 · Hard — bootstrap confidence interval
Given `sample` (30 observations), estimate the **95% confidence interval**
of the mean using bootstrap resampling with `B=10_000` resamples.

Steps:
1. Draw `B` bootstrap samples — each is a resample of size 30 **with replacement**  
   (hint: `rng.integers(0, n, size=(B, n))` gives you `B` rows of indices)
2. Compute the mean of each bootstrap sample → `boot_means` shape `(B,)`
3. Compute `ci_low` and `ci_high` — the 2.5th and 97.5th percentiles of `boot_means`
4. Verify the true mean falls inside the CI


```python
rng = np.random.default_rng(42)
sample = rng.normal(loc=10.0, scale=2.0, size=30)
B = 10_000
n = len(sample)

# Draw B bootstrap samples (each row is one resample with replacement)
idx  = rng.integers(0, n, size=(B, n))          # shape (B, 30)
boot_samples = sample[idx]                      # shape (B, 30)
boot_means = boot_samples.mean(axis=1)          # shape (B,)

ci_low  = np.percentile(boot_means, 2.5)
ci_high = np.percentile(boot_means, 97.5)

```

```python
assert boot_means.shape == (B,)
assert ci_low  < sample.mean() < ci_high, "True mean should be inside the CI"
assert ci_low  > sample.mean() - 3,       "CI lower bound seems too low"
assert ci_high < sample.mean() + 3,       "CI upper bound seems too high"
print(f"✓ Problem 10 passed")
print(f"sample mean : {sample.mean():.4f}")
print(f" 95% CI : [{ci_low:.4f}, {ci_high:.4f}]")
```
